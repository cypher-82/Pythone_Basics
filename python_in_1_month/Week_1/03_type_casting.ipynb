> # **Type Casting in Python**

- ### 1. Implicit Casting

```python
a = 7
b = 8.2
c = a + b
print(c) # Python Automaticaly convert integer into float.
```

- ### 2. Explicit Casting

```python
# string into integer
var_0 = '23'
con = int(var_0)
print(f'{con} : {type(con)}')

# integer into string
var_1 = 456
con_1 = str(var_1)
print(f'{con_1} : {type(con_1)}')

# string into float
var_3 = float("3.6987")
print(f'{var_3} : {type(var_3)}')

# float into string
var_2 = str(2.33)
print(f'{var_2} : {type(var_2)}')

# list into set
my_list = [0, 1, 1, 2, 3, 5, 8, 13]
con_2 = set(my_list)
print(f'{con_2} : {type(con_2)}')

# set into list
my_set = {44, 56, 7, 34, 8, 8, 56}
con_3 = list(my_set)
print(f'{con_3} : {type(con_3)}')
```
