> # **Functions in Python**
   - ## A function is a reusable block of code designed to perform a specific task. You define the logic once and call it whenever you need that task

> ### **What is their role?**

Functions make programs organized, reusable, readable, testable, and easier to maintain. They let
you break a large problem into smaller, understandable pieces.

- **DEFINE** → write the behavior
- **CALL** → execute it
- **INPUT** → give data
- **OUTPUT** → receive a result


- ## 1. WHY FUNCTIONS EXIST
   - If the same logic is needed many times, copying it creates duplication. A function lets you write the logic once and reuse it.

```python
def calculate_total(price, quantity):
    return price * quantity

print(calculate_total(100, 3))
print(calculate_total(260, 2))
```

> **The function name describes a reusable behavior. This makes the program easier to read and easier to
change.**

- ## 2. DEF, NAME, PARENTHESES & INDENTATION
   - Python uses **def** to define a function. The name identifies it; parentheses hold parameters; the colon starts the body; indentation defines which statements belong to it.

> **Defining a function normally does not execute its body. The body runs when you call the function.**

```python
def greet():
    print("Hello!")

greet()
```

- ## 3. CALLING A FUNCTION

> **Think: definition = what function does; call = when it does it.**

```python
def show_message():
    print("Python is Fun!")

show_message()
show_message()
```

- ## 4. PARAMETERS AND ARGUMENTS
   - A **parameter** is a variable in the function definition. An **argument** is the actual value supplied when calling the function

> **name is the parameter. Hammad and Ali are arguments**

```python
def greet(name):
    print("Hello", name)

greet("Hammad Butt")
greet("Ali")
```

- ## 5. RETURN
   - **return** sends a value back to the caller. It is different from print(): print displays; return provides a value the rest of the program can use.

```python
def add(a, b):
    return a + b

result = add(10, 6)
print(result)
    
```

```python
def square(x):
    return x * x

x = square(5)
y = square(x)
print(y)
```

- ## 6. RETURN STOPS THE FUNCTION
   - When return executes, the current function call ends. Code after that return in the same execution is not reached.

```python
def test():
    print("Start")
    return 10
    print("Never runs")

print(test())
```

- ## 7. MULTIPLE RETURN VALUES
   - Python can return several values. They are packaged as a tuple and can be unpacked into separate variables.

```python
def add(a, b):
    return a + b, a - b

x, y = add(7, 3)
print(x)
print(y)
```

- ## 8. DEFAULT PARAMETERS
   - A default parameter is used when the caller does not provide that argument.

> **Default parameters make a function flexible while keeping common calls short.**

```python
def greet(name, message = "Hello"):
    print(message, name)

greet("Hammad")
greet("Hammad", "Welcome")
```
