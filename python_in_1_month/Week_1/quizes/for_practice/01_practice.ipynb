> # **Practive Programs (Loops)**

-  > ## LEVEL 1 - BASIC FOR

```python
# Print 1-10

for i in range(1, 11):
    print(i, end=" ")
```

```python
# Print 10-1

for i in range(10, 0, -1):
    print(i, end=" ")
```

```python
# Print even numbers from 1-20

for i in range(2, 21, 2):
    print(i, end=", ")
```

```python
# Calculate the sum from 1-10

total = 0

for i in range(1, 11):
    total = total + i

print(total)
```

-  > ## LEVEL 2 - FOR + IF

```python
# Print Multiplies of 3

for i in range(3, 100):
    if i % 3 == 0:
        print(i, end=", ")
```

```python
# Count even numbers

for i in range(2, 100):
    if i % 2 == 0:
        print(i, end=", ")
```

```python
# Find numbers Divisible by 5

for i in range(5, 50):
    if i % 5 == 0:
        print(i, end=", ")
```

```python
# Check weather a number is prime.

n = int(input("Enter number : "))

if n < 2:
    print(f"{n} is not prime")
else:
    is_prime = True

    for d in range(2, n):
        if n % d == 0:
            is_prime = False
            break

    if is_prime:
        print(f'{n} is Prime')
    else:
        print(f"{n} is not Prime")
```

-  > ## LEVEL 3 - WHILE

```python
# Print 1-10 with while.

n = 1

while n < 11:
    print(n, end=" ")
    n += 1
```

```python
# Keep asking for a positive number.

n = int(input("Enter number : "))

while n < 0:
    print("Negative integers are not allowed!")
    n = int(input('Enter number : '))

if n > 0:
    print(f'{n} is positive integer.')
```

```python
# Password Validator

password = input("Enter controls password : ")

while password != "123cc":
    print("You entered the wrong pasword!")
    password = input("Enter controls password : ")

if password:
        print(f'{password} is correct controls password.')
```

-  > ## LEVEL 4 - CONTROL

```python
for i in range(1, 100):
    if i == 51:
        break
    elif i == 5:
        continue
    print(i, end=" ")
```

```python
# Find the firt number Divisible by 7

for i in range(1, 30):
    if i % 7 == 0:
        if i == 7:
            continue
        print(i, end=" ")
```

-  > ## LEVEL 5 - NEASTED

```python
# Make a star triangle

for i in range(1, 8):
    for j in range(i):
        print("","*", end=" ")
    print()
```

```python
# Make a square pattern

for i in range(1, 7):
    for j in range(1, 7):
        print("", "*", end=" ")
    print()
```

```python
# Build 5x5 Multiplication Table

for i in range(1, 6):
    for j in range(1, 6):
        print(i * j, end="\t")
    print()
```

-  > ## LEVEL 6 - NUMBER LOGIC

```python
# Factorial

n = int(input("Enter the number"))
fact = 1

while n < 0:
    print("Your enter negative integer which is not allowed!")
    n = int(input("Enter the number : "))

print("----------------------------")

for i in range(1, n + 1):
    fact = fact * i

print(f'The factorial of integer {n} is {fact}')

print("----------------------------")
```

```python
# Reverse a number or string

name = input("Enter any word/name/place/noun etc.. : ")
rev = ""

for char in name:
    rev = char + rev

print(f'The reverse of your word \'{name}\' is \'{rev}\'. Thank You....')
```

```python
# Palindorm cheak

word = input("Enter : ")
word = word.lower()
rev = ""

for char in word:
    rev = char + rev

if word == rev:
    print(f'{word} reverse is \'{rev}\' and is Palindorm')
else:
    print(f'{word} reverse is \'{rev}\' and is not Palindorm')
```
