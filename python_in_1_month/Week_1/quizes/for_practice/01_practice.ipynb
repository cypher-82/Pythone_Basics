> # **Practive Programs (Loops)**

```python
# Print 1-10

for i in range(1, 11):
    print(i, end=" ")
```

```python
# Print 10-1

for i in range(10, 0, -1):
    print(i, end=" ")
```

```python
# Print even numbers from 1-20

for i in range(2, 21, 2):
    print(i, end=", ")
```

```python
# Calculate the sum from 1-10

total = 0

for i in range(1, 11):
    total = total + i

print(total)
```

```python
# Print Multiplies of 3

for i in range(3, 100):
    if i % 3 == 0:
        print(i, end=", ")
```

```python
# Count even numbers

for i in range(2, 100):
    if i % 2 == 0:
        print(i, end=", ")
```

```python
# Find numbers Divisible by 5

for i in range(5, 50):
    if i % 5 == 0:
        print(i, end=", ")
```

```python
# Check weather a number is prime.

n = int(input("Enter number : "))

if n < 2:
    print(f"{n} is not prime")
else:
    is_prime = True

    for d in range(2, n):
        if n % d == 0:
            is_prime = False
            break

    if is_prime:
        print(f'{n} is Prime')
    else:
        print(f"{n} is not Prime")
```

```python
# Print 1-10 with while.

n = 1

while n < 11:
    print(n, end=" ")
    n += 1
```

```python
# Keep asking for a positive number.

n = int(input("Enter number : "))

while n < 0:
    print("Negative integers are not allowed!")
    n = int(input('Enter number : '))

if n > 0:
    print(f'{n} is positive integer.')
```
