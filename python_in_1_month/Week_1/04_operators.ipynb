> # **Operators in Python**

- ### 1. Arithmetic Operators

```python
a, b = 10, 4

print(a + b) # 14 addition
print(a - b) # 6 subtraction
print(a * b) # 40 multiplication
print(a / b) # 2.5 true division 
print(a // b) # 2 florr divsion (decimal cut)
print(a % b) # 2 moduluzzs (reminder)
print(a ** b) # 10000 power (10 to the power 4)
```

- ### 2. Comparison / Relational Operators
   - result always TRUE or FALSE.

```python
a, b = 23, 6

print(a == b) # equals to (False)
print(a != b) # not equals to (True)
print(a > b) # greater then (True)
print(a < b) # less then (False)
print(a >= b) # greater then or equals to (True)
print(a <= b) # less then or equals to (False)
```

- ### 3. Logical Operators

```python
x = True
y = False

print(x and y) # False (Both have to True)
print(x or y) # True (One have be True)
print(not y) # True (Convert False into True and True into False)
```

- ### 4. Assignment Operators

```python
x = 4

x += 6 # x = x + 3 → 7
x -= 2 # x = x - 2 → 2
x *= 8 # x = x * 8 → 32
x /= 2 # x = x / 2 → 2
x //= 6 # x = x // 2 → 0
x %= 5 # x = x % 5 → 1
x **= 2 # x = x ** 2 → 16
```

- ### 5. Identity Operators
   - Check karte hain ke do variables same memory location par hain ya nahi:

```python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a is b) # True
print(a is c) # False
```

- ### 6. Membership Operators
   - Kisi cheez ke andar element hai ya nahi check karte hain:

```python
a = [2, 4, 6, 8, 10, 12]

print(2 in a) # True
print(5 in a) # False
print(7 not in a) # True
```

- ### 7. Bitwise Operators
   - Advanced — binary level pe kaam karta hai

```python
a, b = 5, 3      # binary: 101, 011

print(a & b)     # 1  AND
print(a | b)     # 7  OR
print(a ^ b)     # 6  XOR
print(~a)        # -6 NOT
print(a << 1)    # 10 left shift
print(a >> 1)    # 2  right shift
```
