> # **Data Types in Python**

- ### 1. Numerical Data Types 

```python
a = 18 # int
b = 3.14 # float
c = 3 + 4j # complex

print(a + b + c)
```

- ### 2. Strings (Text) Data Type

```python
name = 'Hammad Butt' # Single quote string
name2  = "C Y P H E R" # Double quote string
name3 = '''Multi Line String
Use full for add pragraphs etc...''' # Triple quote string

print(name)
print(name2)
print(name3)

# Strings are indexed, ordered and immutable data types. Strings are a sequence of characters. Each character in a string has an index number starting from 0.

nikename = 'Cypher'
print(nikename[0]) # C
print(nikename[1]) # y
print(nikename[2]) # p
print(nikename[3:6]) # her
```

- ### 3. Boolean Data Types

```python
is_done = True # Boolean
is_active = False # Boolean

print(f"{type(is_done)} \n{type(is_active)}") # <class 'bool'> <class 'bool'>
```

- ### 4. None

```python
data = None # NoneType
print(f"{type(data)}") # <class 'NoneType'>
```

- ### 5. Collection Data Types

Type        -         Symbol      -      Ordered?    -       Mutable?     -       Duplicate allowed?

List        -          []         -       YES        -        YES         -            YES

Tuple       -          ()         -      YES         -     NO(Fixed)      -           YES

Set         -          {}         -       NO         -        YES         -  NO(Duplicates not allowed)

Dictionary  -     {Key: Value}    -   YES(3.7+)      -       YES          -       YES(Keys are unique)

```python
# List 
my_list = [1, 2, 3, 4, 5, 6, 'Moarning'] # Lists are mutable, meaning you can modify, add, or remove elements after the list has been created
my_list[6] = 'Evening'

# Tuple
my_tuple = (1, 2, 3, 4, 'Good', 5) # Tuples are immutable, meaning once they are created, their contents cannot be changed or altered in any way

# Set
whole_numbers = {0, 1, 2, 3, 4, 5, ...}

# Dictionary
personal_info = {
    'name' : 'Hammad Butt',
    'age' : 20,
    'height' : 178,
    'DOB' : f'{2007}/{12}/{12}',
    'non' : 'Everyone'
}

print(type(my_list))
print(type(my_tuple))
print(type(whole_numbers))
print(type(personal_info))

print(f'{my_tuple[4]} {my_list[6]} {personal_info["non"]}')
```
