> # **Conditional Statments in Python**
   - ## Decision Making

- ## 1. [ if ] Statement -- Most Basic
  
     - Important Points
         1. [ if ] ke baad condition likhte hain jo True ya False mein evaluate hoti hai.
         2. Condition ke baad colon : lazmi hai.
         3. Agle line se indentation (4 spaces ya 1 tab) shuru hoti hai — Python indentation se hi blocks pehchanta hai (curly braces {} nahi hoti jaisa C/Java mein)

     - Common Mistakes
         - if age >= 18: 
         - print("Adult")   # IndentationError — indent nahi kiya

```python
age = 20
if age >= 18:
    print("You are adult.")

print('-----------------------------------------------------')

height = float(input('Your Height? : '))
if height >= 170:
    print('Congatulations You are eligiable to compete!')
```

- ## 2. [ if-else ] -- Two ways

```python
age = 17

if age >= 18:
    print('You are Adult')
else:
    print('You are Minor')
```

- ## 3. [ if-elif-else ] -- Multiple ways

```python
marks = 81

if marks >= 90:
    print('A+')
elif marks >= 80:
    print('A')
elif marks >= 70:
    print('B+')
elif marks >= 60:
    print('B')
elif marks >= 50:
    print('C')
elif marks >= 45:
    print('D')
else:
    print("Fail")
```

  - Kaise kaam karta hai: Python upar se neeche check karta hai, jaise hi koi condition True milti hai, wahi block chalta hai aur baaki saare elif/else skip ho jate hain — chahe wo bhi True hon.

```python
x = 10

if x > 5:
    print("Greater")
elif x > 8:
    print('Again Greater') # Ye kabhi nai chalega, chahe true ho
```

- ## 4. Nested Conditionals (Condition inside Condition)

```python
age = int(input('Enter your age : '))
has_id = input('Is you have ID (y/n) : ')

if has_id == "y":
    has_id = True
elif has_id == "n":
    has_id = False
else:
    print("Please make sure is you have an ID! because,")

if age >= 18:
    if has_id:
        print("Entry Allowed")
    else:
        print("ID Required! ")
else:
    print("Entry not Allowed, Underage!")
```

- ## 5. Logical Operators with Conditions (Compact method)
   - Isko nested karne ki bajaye and/or se ek hi line mein likh sakte    hain — cleaner code.

```python
age = int(input('Enter your age : '))
has_id = input('You have an ID? (y/n) : ')

if has_id == 'y':
    has_id = True
elif has_id == 'n':
    has_id = False
    print('Please make sure you have an ID!')

if age >= 18 and has_id:
    print("Your are adult.")
else:
    print("Your are Minor.")
```

- ## 6. Ternary Operator (One-Line if-else)

```python
age = int(input('Enter your age : '))
has_id = input('You have ID? (y/n) : ')

checking = has_id == True if has_id == 'y' else has_id == False 
status = 'You are Adult.' if age >= 18 and has_id else 'Your are Minor.'

print(status)
```
