>## __Variables in Python__

```python
# These example for Boolean Data Types in python. Boolean values can be either True or False. In this example, we will compare two numbers and print the results of the comparisons.

a = 1.3
b = 4.0

bool_1 = a == b
bool_2 = a > b
bool_3 = a < b
bool_4 = a >= b
bool_5 = a <= b

print(bool_1)
print(bool_2)
print(bool_3)
print(bool_4)
print(bool_5)
```
