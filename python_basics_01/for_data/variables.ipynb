>## __Variables in Python__

### **What is Variables?**
- Variables is the name where you store values.

### **Rules for naming the variable**
> - Only letters, numbers and underscore (_) are use to name the variable.
> - Variable name dose not start with any number use 'num1' instead of '1num'.
> - Case Sensitive → 'Age' and 'age' these two are different vairables.
> - Dont use python reserved key words (if, for, class etc...).
> - Convention: We use snake_case → student_name, total_marks.
> - For constant we use upercase letters like PI = 3.147

```python
name, height, age = "Hammad But", 177, 20
name = 'Cypher'
height = 180
age = 21

a = b = c = 496

print(f"My name is {name}. And my age is {age} years old. My height is {height}cm.")
print(f"The value of a, b, and c is {a}.")

# for constant
PI = 3.14159
print(f"The value of PI is approximately {PI}.")
```
